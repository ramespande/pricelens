# Milestone 1 EDA
Set `DATA_ROOT` to the dataset root before running. This notebook intentionally avoids image pixel processing.

```python
import os
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns
from src.data.loading import load_train, load_test
from src.data.quality import data_quality_report
from src.features.text import extract_text_features
root = os.environ['DATA_ROOT']
train, test = load_train(root), load_test(root)
print(data_quality_report(train))
print(data_quality_report(test))
```

```python
fig_dir=Path('reports/figures'); fig_dir.mkdir(parents=True, exist_ok=True)
fig, axes=plt.subplots(1,2,figsize=(12,4))
sns.histplot(train.price, bins=80, ax=axes[0]); axes[0].set_title('Price')
sns.histplot(train.price.map(__import__('numpy').log1p), bins=80, ax=axes[1]); axes[1].set_title('log1p(price)')
fig.tight_layout(); fig.savefig(fig_dir/'price_distributions.png', dpi=160)
```

```python
features=extract_text_features(train.catalog_content)
features.describe().T
fig, ax=plt.subplots(figsize=(7,4)); sns.histplot(features.char_count, bins=80, ax=ax); ax.set_title('Catalog content character count'); fig.tight_layout(); fig.savefig(fig_dir/'text_length_distribution.png', dpi=160)
train[['sample_id','catalog_content','price']].sample(5, random_state=42)
```
